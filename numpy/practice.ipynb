```python
arr1 = np.array([1, 2, 3])

arr2 = arr1

arr2[0] = 100

arr1, arr2
```

```python
arr1 = np.arange(0, 1, 0.1)

view = arr1.view()

view[0] = 100.0

view, arr1
```

```python
arr2 = np.arange(0, 1, 0.1)

cpy = arr2.copy()

cpy[0] = 100

cpy, arr2
```

```python
cpy.base
```

```python
view.base
```

```python
np.shares_memory(arr1, view)
```

```python
np.shares_memory(arr2, cpy)
```
